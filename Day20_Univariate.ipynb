# Exploratatory Data Analysis(EDA)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('train.csv')
```

```python
df.head(5)
```

# 1. Categorical Data

## a.CountPlot

```python
sns.countplot(x=df['Survived'])
```

## b. Pie Charts

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

# 2. Numerical Data

## a.Histogram

```python
plt.hist(df['Age'])
```

## b.DistPlot

```python
sns.distplot(df['Age'])
```

## c.BoxPlot

```python
sns.boxplot(x=df['Age'])
```

```python
df['Age'].skew()
```
