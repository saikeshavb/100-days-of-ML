```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import scipy
sns.set(rc={"figure.figsize": (5, 3)})
```

```python
tips=pd.read_csv('tips.csv')
tips.head(5)
```

```python
flights=pd.read_csv('flights.csv')
flights.head(5)
```

```python
iris=pd.read_csv('iris.csv')
iris.head(5)
```

```python
titanic=pd.read_csv('titanic.csv')
titanic.head(5)
```

# 1. Scatter Plot

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```

# 2. Bar Plot(Numerical-categorical)

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'])
```

# 3. BoxPlot

```python
sns.boxplot(x=titanic['Sex'],y=titanic['Age'])
```

# 4. DistPlot (Numerical to Categorical)

```python
sns.distplot(x=titanic[titanic['Survived']==0]['Age'],hist=False)
sns.distplot(x=titanic[titanic['Survived']==1]['Age'],hist=False)
```

# 5. HeapMap (Categorical - categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

# 6. ClusterMap (Categorical - categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

# 7. PairPlot 

```python
sns.pairplot(iris,hue='species')
```
